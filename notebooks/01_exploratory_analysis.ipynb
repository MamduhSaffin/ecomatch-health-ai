# 01. Exploratory Data Analysis — EcoMatch Health AI
Reproducible analysis of 104 weeks of synthetic medical consumable demand across 8 verified healthcare facilities (Seed `20261012`, zero patient PII).


```python
import csv, statistics
from pathlib import Path

rows = list(csv.DictReader(open('../data/synthetic/weekly_consumption.csv')))
print(f'Total weekly consumption records: {len(rows)}')
by_sku = {}
for r in rows:
    by_sku.setdefault(r['sku'], []).append(int(r['units_consumed']))
for sku, vals in by_sku.items():
    print(f'{sku}: mean={statistics.mean(vals):.1f}, stdev={statistics.stdev(vals):.1f}')
```
