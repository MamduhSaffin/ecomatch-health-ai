# 02. Chronological Feature Engineering
Constructs strictly causal lag (`t-1`, `t-4`, `t-52`), 4-week rolling mean, linear trend, and 1st/2nd order Fourier seasonal harmonics without future data leakage.


```python
import math
def build_causal_features(series, t):
    lag1 = series[max(0, t-1)]
    lag4 = series[max(0, t-4)]
    lag52 = series[t-52] if t >= 52 else lag4
    roll4 = sum(series[max(0, t-4):t]) / max(1, len(series[max(0, t-4):t]))
    ang = 2 * math.pi * t / 52.0
    return [1.0, t/104.0, lag1, lag4, lag52, roll4, math.cos(ang), math.sin(ang), math.cos(2*ang), math.sin(2*ang)]
print('Causal feature vector dimensionality:', len(build_causal_features([20]*60, 55)))
```
